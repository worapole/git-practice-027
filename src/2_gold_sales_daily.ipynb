```python
from pyspark.sql import functions as F

sales = spark.table("ctl_training_dev.m3.silver_sales")

daily_kpi = (
    sales
    .withColumn("sales_date", F.to_date("dateTime"))
    .groupBy("sales_date")
    .agg(
        F.sum("totalPrice").alias("revenue"),
        F.countDistinct("transactionID").alias("transactions"),
        F.countDistinct("customerID").alias("customers")
    )
)

(
    daily_kpi.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("ctl_training_dev.m3.gold_sales_daily")
)

print("gold_sales_daily created")
```
