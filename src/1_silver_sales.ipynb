```python
from pyspark.sql import functions as F

transactions = spark.table("samples.bakehouse.sales_transactions")
customers = spark.table("samples.bakehouse.sales_customers")
franchises = spark.table("samples.bakehouse.sales_franchises")

silver_sales = (
    transactions.alias("t")
    .join(customers.alias("c"), "customerID", "left")
    .join(franchises.alias("f"), "franchiseID", "left")
    .dropDuplicates(["transactionID"])
)

(
    silver_sales.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("ctl_training_dev.m3.silver_sales")
)

print("silver_sales created")
```
